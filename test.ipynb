Question: Subarray Sum Equals K

Given an integer array nums and an integer k, return the number of contiguous subarrays whose sum equals k.

Examples

nums = [1,1,1], k = 2 → 2

nums = [1,2,3], k = 3 → 2

nums = [3,4,7,2,-3,1,4,2], k = 7 → 4

Constraints (interview version)

1 ≤ n ≤ 2e5

-10^4 ≤ nums[i] ≤ 10^4 (negatives allowed)

-10^9 ≤ k ≤ 10^9

Target time: O(n) (hashing with prefix sums). Brute-force O(n²) is acceptable to discuss first but won’t pass large n.

Your turn (Step 1):
Talk me through your approach and time/space complexity. No code yet—just the idea you’d implement. After that, I’ll ask you to code it.

```python
def subarray_sum(arr, k):
    n = len(arr)
    count = 0
    for i in range(n):
        total = 0
        for j in range(i, n):
            total += arr[i]
            if total == k:
                count += 1
    return count

    
```
