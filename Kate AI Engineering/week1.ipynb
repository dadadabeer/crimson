```python
def sum_numbers(n):
    total = 0
    for i in range(0, n+1, 1):
        total = total + i
    print(total)

sum_numbers(5)
sum_numbers(100)
```

```python
# 0 1 1 2 3 5 8 13 21
def fibonacci(n):
    if n == 1:
        return 0
    if n == 2:
        return 1
    return fibonacci(n-1) + fibonacci(n-2)
```
